```python
import pypsa
import pandas as pd
```

```python
n = pypsa.Network("resources/networks/base_s_40__Co2L0-168h-lv1.25-wcf1.0_2050_2013.nc")
# Calcul du facteur de charge moyen et de la ressource théorique
cf_lw25 = n.generators_t.p_max_pu.loc[:, n.generators.carrier == "onwind"].mean().mean()
cf_lw13 = n.generators_t.p_max_pu.loc[:, n.generators.carrier == "onwind_type2"].mean().mean()

print(f"Facteur de charge moyen brut LW25 : {cf_lw25:.4f}")
print(f"Facteur de charge moyen brut LW13 : {cf_lw13:.4f}")
```
